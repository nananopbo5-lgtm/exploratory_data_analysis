```python
# statistical describtion of the dataset where we dont want longitude and latitude coz this two columns are not useful for our analysis
# or if we want only specific columns, we can use the following code
df[['housing_median_age', 'total_rooms', 'total_bedrooms', 'population', 'households', 'median_income', 'median_house_value']].describe()


# here what we found is mean median quartiles std and 
# other statistical values of the dataset, 
# but we can also find out the stastical describtion of the categorical columns by using include = 'object' in describe() function
# 
```

```python
# now we want to see how a bedroom are concentrated across different housing blocks so
# what is the concentration/distribution of total_bedrooms across housing blocks (AKA we call it as societies in a building)?-->it mean how many properties have 1,2,3+ bedrooms    
# the best visualization for this Q is
# we won't use boxplot coz its only when we want to see the outliers.
# so we'll use histogram to see the distribution of total_bedrooms across housing blocks

ax = sns.histplot(data = df, x = 'total_bedrooms', binwidth = 250)

# data labels
for p in ax.patches:
    ax.text(x = p.get_x() + p.get_width() / 2, y = p.get_height(), s = f'{int(p.get_height())}', ha = 'center', va = 'bottom')

plt.title('Distribution of Total Bedrooms')
plt.xlabel('Total Bedrooms')
plt.ylabel('')
plt.yticks([])

plt.gca().spines[['top', 'right', 'left']].set_visible(False)

plt.show()
```

```python
# or what we will do is for the graph size
plt.figure(figsize = (12, 4))
ax = sns.histplot(data = df, x = 'total_bedrooms', binwidth = 250)

# data labels
for p in ax.patches:
    ax.text(x = p.get_x() + p.get_width() / 2, y = p.get_height(), s = f'{int(p.get_height())}', ha = 'center', va = 'bottom')

plt.title('Distribution of Total Bedrooms')
plt.xlabel('Total Bedrooms')
plt.ylabel('')
plt.yticks([])

plt.gca().spines[['top', 'right', 'left']].set_visible(False)

plt.show()
```

```python
# now impute the missing values in total_bedroom with median value becoz media will make sense more here.
df['total_bedrooms'] = df['total_bedrooms'].fillna(df['total_bedrooms'].median())
# fillna means fill the missing values with median value of total_bedrooms column.
```

```python
df.isna().sum()
# now we'll notice the no's became 0, which means we have successfully imputed the missing values in total_bedrooms column with median value.
```

```python
# how are housing values distributed across california?
plt.figure(figsize = (12, 4))
ax = sns.histplot(data = df, x = 'median_house_value', binwidth = 50000) 

# data labels
for p in ax.patches:
    ax.text(x = p.get_x() + p.get_width() / 2, y = p.get_height(), s = f'{int(p.get_height())}', ha = 'center', va = 'bottom')

plt.title('Distribution of Median House Values')
plt.xlabel('Median House Value')
plt.ylabel('')
plt.yticks([])

plt.gca().spines[['top', 'right', 'left']].set_visible(False)

plt.show()

# the results will show us that majority of houses are below the range of 250,000$ and that covers your majority of the data.
```

```python
# how many housing complexes fall in each_proximity category?
plt.figure(figsize = (8, 5))

ax = sns.countplot(data = df, x = 'ocean_proximity', order = df['ocean_proximity'].value_counts().index)

# data labels
for p in ax.patches:
    ax.text(x = p.get_x() + p.get_width() / 2, y = p.get_height(), s = f'{int(p.get_height())}', ha = 'center', va = 'bottom')

plt.title('Count of Housing Complexes by Ocean Proximity')
plt.xlabel('Ocean Proximity')
plt.ylabel('')
plt.yticks([])

plt.gca().spines[['top', 'right', 'left']].set_visible(False)

plt.show()

# so in result you will notice that the majority of houses are less than 1H ocean
```
